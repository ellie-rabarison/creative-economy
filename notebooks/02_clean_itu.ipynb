```python
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/itu_mobile_coverage_data.csv")
```

```python
df.head()
```

```python
df=df.rename(columns={"AreaName": "entity", "AreaCode": "iso3", "Year": "year", "Unit": "unit", "Value": "value"})
```

```python
df.head()
```

Only keep 4G (i271GA)

```python
df=df[df["SeriesCode"]=="i271GA"]
df["SeriesCode"].unique()
```

```python
df["iso3"].unique()
```

```python
keep_iso3 = ["MDG", "MUS", "SYC", "COM", "WLD", "AFR", "LDC", "SIDS"]
df = df[df["iso3"].isin(keep_iso3)]
df["iso3"].unique()
```

```python
islands = ["MDG", "MUS", "SYC", "COM"]
df = df[df["iso3"].isin(islands) | (df["year"] <= 2024)]
```

ITU group aggregates kept up to 2024; 2025 values are largely projected, since most countries had not yet reported. 2024 is the latest year common to all four islands.

```python
df = df[df["year"].between(2015, 2026)]
df.shape
```

```python
df["indicator"] = "lte_coverage"
df["element"] = "Infrastructure"
df["unit"] = "% of population covered by 4G/LTE"
df["source"] = "ITU"
```

```python
df["note"] = df["Source"]
df.loc[df["Notes"].notna() & (df["Notes"] != df["Source"]), "note"] = df["Source"] + " | " + df["Notes"]
```

```python
df = df[["iso3","entity","indicator","element","year","value","unit","source","note"]]
df.head()
```

```python
df = df.sort_values(["indicator", "iso3", "year"])
```

```python
df.head()
```

```python
df[(df["year"] == 2023)]
```

```python
df = df[df["year"].between(2015, 2026)]
df.shape
```

```python
df.to_csv("../data/processed/itu_data.csv", index=False)
```
